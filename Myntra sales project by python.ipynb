```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Mytra sales dashboard.csv",header=1)
```

```python
# what is the total revenue by the company?
Total_revenue = df["Total_amount"].sum()
print(Total_revenue)
```

```python
# what is the total profit generated by the company?
Total_profit = df["total_profit"].sum()
print(Total_profit)
```

```python
# what is the overall profit margin percentage?
Total_revenue = df["Total_amount"].sum()
Total_profit = df["total_profit"].sum()
profit_margin = (Total_profit/Total_revenue)*100
print("overall profit margin:",round(profit_margin,2),"%")
```

```python
# Which category contribute the most total profit ?
Top_category = df.groupby("Category")["total_profit"].sum().sort_values(ascending=False).head(1)
print(Top_category)
```

```python
# which product have the highest sales quantity?
Top_product_sales = df.groupby("Product_name")["sales"].sum().sort_values(ascending=False).head(1)
print(Top_product_sales)
```

```python
# Which product have low sales performance?
Low_product_sales = df.groupby("Product_name")["sales"].sum().sort_values(ascending=True).head(1)
print(Low_product_sales)
```

```python
# What percentage of total revenue comes from each category?
Category_revenue = df.groupby("Category")["Total_amount"].sum()
Total_revenue = df["Total_amount"].sum()
Category_percentage = (Category_revenue/Total_revenue)*100
print(Category_percentage.round(2))
```

```python
# which product generated  highest profit ?
Highest_product = df.groupby("Product_name")["total_profit"].sum().sort_values(ascending=False).head(1)
print(Highest_product)
```

```python
# What is the average sales order value?
avg_sales = df.groupby("Product_name")["sales"].mean()
print(avg_sales)
```

```python
# What is the year sales revenue?
year_growth = df.groupby("Year")["sales"].sum().sort_values(ascending=True)
print(year_growth)
```

```python
# what is the year over year sales growth?
Year_sales = df.groupby("Year")["Total_amount"].sum().sort_index()
Year_growth = Year_sales.pct_change()*100
print(year_growth.round(2))
```

```python
 # Which category has the highest average  selling price?
Avg_selling_price = df.groupby("Category")["Selling_price"].mean().sort_values(ascending=True)
plt.pie(
    Avg_selling_price.values,
    labels=Avg_selling_price.index,
    autopct="%1.1f%%",
    startangle=45
)
plt.title("Category wise avg selling price")
plt.show()
```

```python
# Retrive the year wise profit revenue using bar graph?
Yearly_profit_revenue = df.groupby("Year")["total_profit"].sum().sort_values(ascending=True)
sns.barplot(
    x=Yearly_profit_revenue.index,
    y=Yearly_profit_revenue.values,
    hue = Yearly_profit_revenue.index,
    palette="viridis",
    legend=False
)
plt.title("Yearly profit revenue")
plt.xlabel("Year")
plt.ylabel("profit revenue")
plt.show()
```

```python
# Which state generated the highest proit revenue using line chart?
state_profit_revenue = df.groupby("state")["total_profit"].sum().sort_values(ascending=True).head(5)
plt.plot(state_profit_revenue,
         state_profit_revenue,
         marker="o",
         linestyle="dashed"
        )
plt.title("State wise profit revenue")
plt.xlabel("State")
plt.ylabel("profit revenue")
plt.figure(figsize=(10,5))
plt.show()
```

```python
# Display the Top 5 city sales revenue using bargraph
Top5_city = df.groupby("city")["sales"].sum().sort_values(ascending=True).head(5)
sns.barplot(x=Top5_city.index,y=Top5_city.values,hue = Top5_city.index)
plt.title("City wise sales revenue")
plt.xlabel("city")
plt.ylabel("sales")
plt.show()
```

```python
# display the top 5 city profit revenue?
Top5_profit_revenue = df.groupby("city")["total_profit"].sum().sort_values(ascending=False).head(5)
sns.barplot(x=Top5_profit_revenue.index,
            y=Top5_profit_revenue.values,
            hue = Top5_profit_revenue
           )
plt.title("City wise profit revenue")
plt.xlabel("city")
plt.ylabel("profit revenue")
plt.figure(figsize=(12,6))
plt.show()
print(Top5_profit_revenue.head(5))
```

```python
# display the top 5 product sales revenue
Top5_product_name = df.groupby("Product_name")["sales"].sum().sort_values(ascending=True).head(5)
sns.barplot(x=Top5_product_name.index,
            y=Top5_product_name.values,
            hue = Top5_product_name.index
           )
plt.title("Top 5 product sales revenue")
plt.xlabel("Product name")
plt.ylabel("sales revenue")
plt.show()
```

```python
# Display the top 5 product name by the profit revenue?
Top5_product_profit = df.groupby("Product_name")["total_profit"].sum().sort_values(ascending=True).head(5)
sns.barplot(x=Top5_product_profit,
            y=Top5_product_profit,
            hue= Top5_product_profit.index
           )
plt.title("Product wise profit revenue")
plt.xlabel("Product")
plt.ylabel("Profit revenue")
plt.show()
```

```python
# display the top 5 monthly profit revenue?
Top5_month = df.groupby("Month")["total_profit"].sum().sort_values(ascending=True).head(5)
plt.plot(Top5_month,
         Top5_month,
         marker="o",
         linestyle="dashed"
        )

plt.title("TOP 5 Month profit revenue")
plt.xlabel("Month")
plt.ylabel("profit revenue")
plt.show()
```

```python
# Display the total revenue by the monthly ?
Top5_month = df.groupby("Month")["Total_amount"].sum().sort_values(ascending=True).head(5)
sns.barplot(x=Top5_month.index,
            y=Top5_month.values,
            hue = Top5_month
           )
plt.title("Top 5 Month Total revenue")
plt.xlabel("Month")
plt.ylabel("profit revenue")
plt.show()
```

```python
# display the gender wise profit revenue?
Profit_revenue = df.groupby("Gender")["total_profit"].sum()
sns.barplot(x=Profit_revenue.index,
            y=Profit_revenue.values,
            hue = Profit_revenue.index,
           )
plt.title("Gender wise sales revenue")
plt.xlabel("Gender")
plt.ylabel("Profit")
plt.show()
```
